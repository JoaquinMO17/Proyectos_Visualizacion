```python
import pandas as pd

# === Load dataset ===
df = pd.read_csv("Crimes_-_2001_to_Present_20251023.csv")
```

```python
import pandas as pd
import plotly.express as px

# --- Top 15 locations ---
top_locations = df["Location Description"].value_counts().nlargest(15).index
df_loc = df[df["Location Description"].isin(top_locations)]

# --- Top 10 crime types ---
top_crimes = df["Primary Type"].value_counts().nlargest(10).index
df_loc = df_loc[df_loc["Primary Type"].isin(top_crimes)]

# --- Create pivot table for heatmap ---
heat_data = (
    df_loc.groupby(["Location Description", "Primary Type"])
    .size()
    .reset_index(name="Count")
    .pivot(index="Location Description", columns="Primary Type", values="Count")
    .fillna(0)
)

# --- Create heatmap ---
fig = px.imshow(
    heat_data,
    labels=dict(x="Crime Type", y="Location", color="Number of Crimes"),
    x=heat_data.columns,
    y=heat_data.index,
    color_continuous_scale="YlOrRd",
    title="Heatmap of Crime Types Across Top 15 Locations in Chicago (2020–2025)"
)

# --- Customize layout ---
fig.update_layout(
    title_font_size=22,
    title_font_color="black",
    xaxis_title="Crime Type",
    yaxis_title="Location",
    paper_bgcolor="white",
    plot_bgcolor="white",
    margin=dict(t=80, l=80, r=50, b=80)
)

fig.show()
```

# Insights
**1. Concentration of Specific Crimes in Certain Locations**

The heatmap reveals strong correlations between particular crime types and specific locations. For example, thefts and robberies are highly concentrated in areas like streets and gas stations.

**2. Public Spaces Show Higher Crime Diversity**

Locations such as streets and sidewalks display a broad distribution of different crime types. This suggests that open public areas are more exposed to a variety of offenses, likely due to higher population density and activity levels.

**3. Strategic Value for Law Enforcement and Policy**

Identifying these location-crime relationships helps prioritize surveillance and resource allocation. For instance, law enforcement can intensify patrols or install surveillance systems in areas where violent or property crimes cluster.

**4. Importance of Temporal and Spatial Integration**

While this heatmap focuses on location and type, combining it with time-based analysis could uncover even deeper patterns—like crime peaks during specific times of day or months.
